```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

Dim_User

```python
df_user = spark.readStream.format("cloudFiles")\
            .option("cloudFiles.format", "parquet")\
            .option("cloudFiles.schemaLocation", "abfss://silver@storagedeproject1.dfs.core.windows.net/DimUser/schema")\
            .option("schemaEvolutionMode", "addNewColumns")\
            .load("abfss://bronze@storagedeproject1.dfs.core.windows.net/DimUser")
```

```python
df_user.drop("_rescued_data")
```

```python
df_user.writeStream.format("delta")\
    .option("checkpointLocation", "abfss://silver@storagedeproject1.dfs.core.windows.net/DimUser/checkpoint")\
    .option("mergeSchema", "true")\
    .trigger(once=True)\
    .option("path", "abfss://silver@storagedeproject1.dfs.core.windows.net/DimUser/data")\
    .toTable("spotify_databricks.silver.Dim_User")
```
